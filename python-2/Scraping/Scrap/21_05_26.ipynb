```python
import numpy as np
import pandas as pd
from bs4 import BeautifulSoup
with open('vishal_new.html','r',encoding='UTF-8') as f:
    html = f.read()

soup = BeautifulSoup(html, 'html.parser')
```

```python

box = soup.find_all('div',class_='_3pLy-c row')
len(box)
```

```python
data=  {'mobilename':[],'star':[],'rating':[],'review':[],'display':[],'camera':[],'new_price':[],'old_price':[],'discount':[]}
```

```python
for i in box:
    data['mobilename'].append(i.find('div',class_='_4rR01T').text.strip())
    data['star'].append(float(i.find('div',class_='_3LWZlK').text.strip()))
    data['rating'].append(int(i.find('span',class_='_2_R_DZ').text.split()[0].replace(',','')))
    data['review'].append(int(i.find('span',class_='_2_R_DZ').text.split()[3].replace(',','')))
    data['display'].append(i.find('ul',class_='_1xgFaf').find_all('li')[1].text.strip())
    data['camera'].append(i.find('ul',class_='_1xgFaf').find_all('li')[2].text.strip())
    data['new_price'].append(float(i.find('div',class_='col col-5-12 nlI3QM').find('div',class_='_30jeq3 _1_WHN1').text.strip().replace('₹','').replace(',','')))
    try:
        data['old_price'].append(float(i.find('div',class_='col col-5-12 nlI3QM').find('div',class_='_3I9_wc _27UcVY').text.strip().replace('₹','').replace(',','')))
    except:
        data['old_price'].append(np.nan)
    try: 
        data['discount'].append(float(i.find('div',class_='_3Ay6Sb').text.split('%')[0]))
    except:
        data['discount'].append(np.nan)
len(data['mobilename'])
```

```python
box[1].find('div',class_='_3Ay6Sb').text.split('%')[0]
```

```python
df = pd.DataFrame(data,columns=['mobilename','star','rating','review','display','camera','new_price','old_price','discount'])
```

```python

with open('List of state and union territory capitals in India - Wikipedia (5_28_2025 8：46：50 AM).html','r',encoding='utf-8') as f:
    html = f.read()
soup = BeautifulSoup(html, 'html.parser')
table = soup.find('table',class_='wikitable sortable jquery-tablesorter')
col=[]
head = table.find('thead')
for i in head.find_all('th'):
    col.append(i.text.strip())
col[5]=col[5].split('\n')[0]
```

```python
body = table.find('tbody')
d={col[0]:[],col[1]:[],col[2]:[],col[3]:[],col[4]:[],col[5]:[]}

for i in body.find_all('tr'):
    l = i.find_all('td')
    d[col[0]].append(l[0].text.strip())
    d[col[1]].append(l[1].text.strip())
    d[col[2]].append(l[2].text.strip())
    d[col[3]].append(l[3].text.strip())
    d[col[4]].append(l[4].text.strip())
    d[col[5]].append(l[5].text.strip())
df = pd.DataFrame(d)
df
```

```python
with open('140 Essential Action Movies To Watch Now ｜ Rotten Tomatoes (5_28_2025 8：24：58 AM).html','r',encoding='utf-8') as f:
    html = f.read()
soup = BeautifulSoup(html,'html.parser')

moves= soup.find_all('div',class_='row countdown-item')
```

```python
df = pd.DataFrame(d,columns=[''])
```
