```python
url ='https://webscraper.io/test-sites/e-commerce/allinone/computers/tablets'
res = requests.get(url)
soup = BeautifulSoup(res.text,'html.parser')
all_Card = soup.find_all('div',class_='product-wrapper card-body')
len(all_Card)




```

```python
import pandas as pd
d={'name':[],'price':[],'reivews':[]}
for i in all_Card:
    d['name'].append(i.find('a').text.strip())
    d['price'].append(i.find('span').text.strip())
    d['reivews'].append(i.find('p',class_='review-count float-end').text.strip())

df = pd.DataFrame(d)
df
```
