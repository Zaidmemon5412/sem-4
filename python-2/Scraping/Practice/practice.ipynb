```python
import requests 
import pandas as pd 
import numpy as np
from bs4 import BeautifulSoup
```

```python
url = 'https://news.google.com/topics/?hl=en-IN&gl=IN&ceid=IN%3Aen'
response = requests.get(url)
print(response.status_code)
soup = BeautifulSoup(response.text)
```

```python
box = soup.find_all('div',class_='LU3Rqb')
len(box)
```

```python
print(box[1].find('a',class_='gPFEn').text.strip())
```

```python
print(box[0].find('div',class_='MCAGUe').find('div',class_='a7P8l')
      .find('div',class_='vr1PYe').text.strip())
```

```python
d={'News':[],'Publisher':[]}
for i in box:
    d['News'].append(i.find('a',class_='gPFEn').text.strip())
    d['Publisher'].append(i.find('div',class_='MCAGUe').find('div',class_='a7P8l')
      .find('div',class_='vr1PYe').text.strip())
    
df = pd.DataFrame(d,columns=['News','Publisher'])
df
```
