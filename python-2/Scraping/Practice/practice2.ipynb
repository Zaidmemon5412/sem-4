```python
import requests
import pandas as pd
import numpy as np
from bs4 import BeautifulSoup
```

```python
with open('../Scrap/140 Essential Action Movies To Watch Now ｜ Rotten Tomatoes (5_28_2025 8：24：58 AM).html',
          'r',encoding='UTF-8') as f:
    html= f.read()
    
```

```python
soup = BeautifulSoup(html,'html.parser')
```

```python
box = soup.find_all('div',class_='row countdown-item')
len(box)
```

```python
data = {'movename':[],'score':[],'img_url':[],'directer':[]}
```

```python
box[0].find('div',class_='row countdown-item-details').find('div',class_='info director').find('a').text
```

```python
for i in box:
    data['movename'].append(i.find('div',class_='col-sm-20 col-full-xs').find('a').text)
    data['score'].append(int(i.find('div',class_='col-sm-20 col-full-xs').find('span',class_='tMeterScore').text.split('%')[0]))
    data['img_url'].append(i.find('div',class_='col-sm-6 col-full-xs').find('img',class_='article_poster').get('src'))
    data['directer'].append(i.find('div',class_='row countdown-item-details').find('div',class_='info director').find('a').text)
```

```python
df = pd.DataFrame(data)
df
```

```python
with open('../Scrap/Certified used Mercedes-Benz for sale (1).html','r',encoding='UTF-8') as f:
    html = f.read()
```

```python

soup = BeautifulSoup(html,'html.parser')
box = soup.find_all('div',class_='vehicle-details')
len(box)
```

```python
data = {'Name':[],'year':[],'mileage':[],'price':[],'rating':[],'reviews':[],'dealer_name':[]}
```

```python
float(box[0].find('div',class_='vehicle-dealer').find('div','sds-rating').find('span',class_='sds-rating__link sds-button-link').text.split(' ')[0].replace('(',''))
```

```python
for i in box:
    data['Name'].append(i.find('h2',class_='title').text[5:]) 
    data['year'].append(i.find('h2',class_='title').text[0:4])
    data['mileage'].append(i.find('div',class_='mileage').text.split(' ')[0].replace(',',''))
    data['price'].append(i.find('div',class_='price-section price-section-vehicle-card').find('span').text.replace('$','').replace(',',''))
    try:
        data['rating'].append(float(i.find('div',class_='vehicle-dealer').find('div','sds-rating').find('span').text))
    except:
        data['rating'].append(np.nan)
    try:
        data['reviews'].append(float(i.find('div',class_='vehicle-dealer').find('div','sds-rating').find('span',class_='sds-rating__link sds-button-link').text.split(' ')[0].replace('(','')))
    except:
        data['reviews'].append(np.nan)
    
    
    data['dealer_name'].append(i.find('div',class_='vehicle-dealer').find('div','dealer-name').text.strip())
```

```python
df = pd.DataFrame(data)
df
```
